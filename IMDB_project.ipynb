```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('./drive/MyDrive/7-IMDB-Movie-Data.csv')
data=data.drop_duplicates()
data
```

```python
print(f'There are {data.shape[0]} rows')
print(f'There are {data.shape[1]} columns')
```

```python
data["Revenue (Millions)"]=pd.to_numeric(data["Revenue (Millions)"], errors="coerce")
data["Revenue (Millions)"]=data["Revenue (Millions)"].fillna(0).astype("int64")
data["Metascore"]=pd.to_numeric(data["Metascore"], errors="coerce")
data["Metascore"]=data["Metascore"].fillna(0).astype("int64")
data.info()
```

```python
data1=data.drop(columns='Rank')
data2=data1.drop(columns='Year')
data2.describe()
```

```python
data1=data.sort_values(by="Rating",ascending=False).head(10)
print("These movies have the best rating:")
data1.loc[:, ["Title", "Rating", "Rank","Director"]]
```

```python
data1=data.sort_values(by="Revenue (Millions)",ascending=False).head(10)
print("These movies have the most revenue:")
data1.loc[:, ["Title", "Revenue (Millions)", "Rank"]]
```

```python
data1=data.sort_values(by="Metascore",ascending=False).head(10)
print("These movies have the best score:")
data1.loc[:, ["Title", "Metascore", "Rank"]]
```

```python
data1=data.sort_values(by="Runtime (Minutes)",ascending=False).head(10)
print("These movies have the biggest runtime:")
data1.loc[:, ["Title", "Runtime (Minutes)", "Rank"]]
```

```python
data1=data.sort_values(by="Votes",ascending=False).head(10)
print("These movies have the most votes:")
data1.loc[:, ["Title", "Votes", "Rank"]]
```

```python
data1=data.sort_values(by="Runtime (Minutes)",ascending=False).head(5)
data1
```

```python
data2=data[data['Runtime (Minutes)']<180]
data2
```

```python
print(f'The average rating of all movies with runtime over 180 minutes is {data1.Rating.mean()}')
print(f'The average rating of all movies with runtime less than 180 minutes is {data2.Rating.mean()}')
print('\nHence, the average rating of longer movies is higher than the average rating of shorter ones.')
```

```python
print(f'The average number of votes for all movies with runtime over 180 minutes is {data1.Votes.mean()}')
print(f'The average number of votes for all movies with runtime less than 180 minutes is {data2.Votes.mean()}')
print('\nHence, the average number of votes per longer movies is higher than the average number of votes per shorter ones. \nSo, audience reception is also more when it comes to longer movies.')
```

```python
print(f'The average revenue per movie, for movies with runtime over 180 minutes is {data1['Revenue (Millions)'].mean()} millions')
print(f'The average revenue per movie, for movies with runtime less than 180 minutes is {data2['Revenue (Millions)'].mean()} millions')
print('\nOn average, loneger movies get less revenue than shorter movies.')
```

```python
print(f'The average metascore of all movies with runtime over 180 minutes is {data1.Metascore.mean()}')
print(f'The average metascore of all movies with runtime less than 180 minutes is {data2.Metascore.mean()}')
print('\nHence, the average metascore of longer movies is much higher than the average metascore of shorter ones.')
```

```python
data4=data.groupby("Director", as_index=False).agg({"Rating": "mean"}).rename(columns={"Rating": "Average Rating"})
print(data4)
```
